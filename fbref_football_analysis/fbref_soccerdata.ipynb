### Overview: 

The code in this notebook demonstrates how to use the `soccerdata` library to fetch football data from FBref. It initializes the FBref object for the English Premier League across multiple seasons, retrieves team match statistics for Chelsea, fetches lineups for a specific match, and obtains event data for that match.

To explore the `soccerdata` library further, visit https://soccerdata.readthedocs.io/en/latest/datasources/FBref.html

### Create an instance of FBRef


Necessary data for the league and seasons specified will be downloaded and cached locally in ``~/soccerdata/data/FBref``

Each call from this class will provide a ``pd.DataFrames`` object for data from http://fbref.com.

This data can then be saved to a database or file for further analysis

```python
fbref = sd.FBref(leagues="ENG-Premier League", seasons=[20, 21, 22, 23, 24, 25])
```

### Get match statistics for a specific team

Here I am getting the match statistics for Chelsea in the Premier League for the seasons 2020-2026.  The `stat_type` parameter is set to "schedule" to get the match schedule and results for the team.

We get details such as:
- the opponent and match result
- numbers of goals scored
- kick-off time and date of the match
- ball possesion of the team in question

In the ``match_report`` field, we have entries of the form `/en/matches/{match_id}/{home team}-{away team}-{match date}-{league}`. From this, can extract the match id which can then be used to extract further statistics from a given match such as the lineups and match events.

For the first entry in the data frame (Brighton-Chelsea), we have `match_id = "9d7641eb"`. We use this to extract lineups and match events from this game.

```python
team_match_stats = fbref.read_team_match_stats(stat_type="schedule", team="Chelsea")
team_match_stats.head()
```

#### Lineups

In the lineup data, we get the names of the players named in the lineup on the day, whether the player started the match or was a substitute, and how many minutes they played.

For example `Ben White` was a starter for Brighton and was played 90 minutes of the game

```python
# lineups
match_lineups = fbref.read_lineup(match_id = "9d7641eb")
match_lineups.head()
```

### Match Events

We extract match events which include:
- goal scored, the player who scored and assited the goal
- substitution, the player coming off and on
- player carded


```python
events = fbref.read_events(match_id=["9d7641eb"])
events.head()
```

The three DataFrames in this notebook can form a relational database with the match ID extracted from `match_report`, such as `9d7641eb`, as the canonical unique key joining the three databases.

1. **`matches`**: one row per match. Store `match_id`, `league`, `season`, `game`, date, home team, away team, final score, and the original `match_report` URL.
2. **`lineups`**: one row per player selected for a match. Store `match_id`, team, player, starter status, position, and minutes played.
3. **`events`**: one row per event. Store `match_id`, team, minute, score, event type, `player1`, and `player2`.

With our database in hand, which can be used to perform further analysis such as:
- team form by season, 
- home versus away performance, 
- goals scored and conceded, 
- possession and formation comparisons, 
- starting-lineup stability, player minutes and substitute usage, 
- the relationship between substitutions and results, 
- goal timing, scoring responses, assists, cards, and event patterns by team or player. 

It can also support predictive features such as recent form, average possession, lineup continuity, and early-goal frequency, provided that features are calculated only from matches before the prediction being evaluated.

Key points to consider:
- `events`: generate `event_id` as the primary key, with `match_id` as a foreign key. Include an event sequence number because several events can occur in the same minute.
